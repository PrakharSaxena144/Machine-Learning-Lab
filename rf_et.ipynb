```python
def accuracy(y_true, y_pred):
    return np.sum(y_true == y_pred) / len(y_true)

def f1_score_macro(y_true, y_pred):
    classes = np.unique(y_true)
    f1s = []
    for c in classes:
        tp = np.sum((y_pred == c) & (y_true == c))
        fp = np.sum((y_pred == c) & (y_true != c))
        fn = np.sum((y_pred != c) & (y_true == c))
        if tp == 0 and fp == 0 and fn == 0:
            f1s.append(0)
            continue
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        if precision + recall == 0:
            f1s.append(0)
        else:
            f1s.append(2 * precision * recall / (precision + recall))
    return np.mean(f1s)
```

```python
def gini(y):
    classes = np.unique(y)
    impurity = 1
    for c in classes:
        p = np.sum(y == c) / len(y)
        impurity -= p**2
    return impurity

def entropy(y):
    classes = np.unique(y)
    ent = 0
    for c in classes:
        p = np.sum(y == c) / len(y)
        if p > 0:
            ent -= p * np.log2(p)
    return ent

def split_dataset(X, y, feature, threshold):
    left_idx = X[:, feature] <= threshold
    right_idx = X[:, feature] > threshold
    return X[left_idx], y[left_idx], X[right_idx], y[right_idx]
```

```python
def custom_kfold(X, y, k=5, shuffle=True, random_state=None):
    n_samples = len(X)
    indices = np.arange(n_samples)

    if shuffle:
        rng = np.random.RandomState(random_state)
        rng.shuffle(indices)

    fold_sizes = np.full(k, n_samples // k, dtype=int)
    fold_sizes[: n_samples % k] += 1

    current = 0
    for fold_size in fold_sizes:
        test_idx = indices[current:current + fold_size]
        train_idx = np.concatenate([indices[:current], indices[current + fold_size:]])
        yield train_idx, test_idx
        current += fold_size


def evaluate_model(model, X, y, k=5):
    accs, f1s = [], []
    for train_idx, test_idx in custom_kfold(X, y, k=k, shuffle=True, random_state=42):
        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y[train_idx], y[test_idx]

        model.fit(X_train, y_train)
        preds = model.predict(X_test)

        accs.append(accuracy(y_test, preds))
        f1s.append(f1_score_macro(y_test, preds))

    return np.mean(accs), np.mean(f1s)
```

```python
class DecisionTree:
    def __init__(self, max_depth=None, criterion="gini", n_features=None, random_split=False):
        self.max_depth = max_depth
        self.criterion = criterion
        self.n_features = n_features
        self.random_split = random_split
        self.tree = None

    def fit(self, X, y, depth=0):
        if len(y) == 0:
            return None

        if len(np.unique(y)) == 1:
            return np.unique(y)[0]

        # Depth limit
        if self.max_depth is not None and depth >= self.max_depth:
            return Counter(y).most_common(1)[0][0]

        # Random feature subset
        n_feats = X.shape[1] if not self.n_features else self.n_features
        features = random.sample(range(X.shape[1]), n_feats)

        best_feat, best_thresh = None, None
        if self.random_split:  # Extra Trees
            feat = random.choice(features)
            thresh = random.choice(np.unique(X[:, feat]))
            best_feat, best_thresh = feat, thresh
        else:  # Random Forest / Decision Tree
            best_score = 1e9
            for feat in features:
                thresholds = np.unique(X[:, feat])
                for thresh in thresholds:
                    X_left, y_left, X_right, y_right = split_dataset(X, y, feat, thresh)
                    if len(y_left) == 0 or len(y_right) == 0:
                        continue
                    p = len(y_left) / len(y)
                    if self.criterion == "gini":
                        score = p * gini(y_left) + (1 - p) * gini(y_right)
                    else:
                        score = p * entropy(y_left) + (1 - p) * entropy(y_right)
                    if score < best_score:
                        best_feat, best_thresh, best_score = feat, thresh, score

        if best_feat is None:
            return Counter(y).most_common(1)[0][0]
        X_left, y_left, X_right, y_right = split_dataset(X, y, best_feat, best_thresh)

        # Prevent empty split crash
        if len(y_left) == 0 or len(y_right) == 0:
            return Counter(y).most_common(1)[0][0]

        left = self.fit(X_left, y_left, depth+1)
        right = self.fit(X_right, y_right, depth+1)
        return (best_feat, best_thresh, left, right)

    def train(self, X, y):
        self.tree = self.fit(X, y)

    def predict_one(self, x, tree):
        if tree is None:
            return 0  
        if not isinstance(tree, tuple):
            return tree
        feat, thresh, left, right = tree
        if x[feat] <= thresh:
            return self.predict_one(x, left)
        else:
            return self.predict_one(x, right)

    def predict(self, X):
        return np.array([self.predict_one(x, self.tree) for x in X])
```

```python
class RandomForest:
    def __init__(self, n_trees=10, max_depth=None, criterion="gini", n_features=None, extra=False):
        self.n_trees = n_trees
        self.max_depth = max_depth
        self.criterion = criterion
        self.n_features = n_features
        self.extra = extra
        self.trees = []

    def bootstrap(self, X, y):
        idx = np.random.choice(len(X), len(X), replace=True)
        return X[idx], y[idx]

    def fit(self, X, y):
        self.trees = []
        for _ in range(self.n_trees):
            Xb, yb = self.bootstrap(X, y)
            tree = DecisionTree(max_depth=self.max_depth,
                                criterion=self.criterion,
                                n_features=self.n_features,
                                random_split=self.extra)
            tree.train(Xb, yb)
            self.trees.append(tree)

    def predict(self, X):
        preds = np.array([tree.predict(X) for tree in self.trees])
        final_preds = []
        for i in range(X.shape[0]):
            votes = preds[:, i]
            final_preds.append(Counter(votes).most_common(1)[0][0])
        return np.array(final_preds)
```

```python
n_trees_list = [10, 15, 20]
depths = [3, 5, None]
criteria = ["gini", "entropy"]

results = {"RF": [], "ET": []}

for n_trees in n_trees_list:
    for depth in depths:
        for crit in criteria:
            rf = RandomForest(n_trees=n_trees, max_depth=depth, criterion=crit,
                              n_features=int(np.sqrt(X.shape[1])), extra=False)
            et = RandomForest(n_trees=n_trees, max_depth=depth, criterion=crit,
                              n_features=int(np.sqrt(X.shape[1])), extra=True)
            acc_rf, f1_rf = evaluate_model(rf, X, y)
            acc_et, f1_et = evaluate_model(et, X, y)
            results["RF"].append((n_trees, depth, crit, acc_rf, f1_rf))
            results["ET"].append((n_trees, depth, crit, acc_et, f1_et))
```

```python
plt.figure(figsize=(12,6))
for clf, color in zip(["RF","ET"], ["blue","green"]):
    xs = [r[0] for r in results[clf]]
    ys = [r[3] for r in results[clf]]
    plt.plot(xs, ys, "o-", label=f"{clf} Accuracy", color=color)
plt.xlabel("Number of Trees")
plt.ylabel("Accuracy")
plt.title("Random Forest vs Extra Trees - Accuracy")
plt.legend()
plt.show()

plt.figure(figsize=(12,6))
for clf, color in zip(["RF","ET"], ["red","orange"]):
    xs = [r[0] for r in results[clf]]
    ys = [r[4] for r in results[clf]]
    plt.plot(xs, ys, "o-", label=f"{clf} F1-score", color=color)
plt.xlabel("Number of Trees")
plt.ylabel("F1-score")
plt.title("Random Forest vs Extra Trees - F1 Score")
plt.legend()
plt.show()
```
